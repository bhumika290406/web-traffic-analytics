```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded successfully!")
import os

os.listdir()
df = pd.read_csv('/content/web_traffic_analytics_v2.csv')

df.head()
print("Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
df.head(10)
df.info()
print("StartedRegistration values:")
print(df['StartedRegistration'].value_counts())

print("\nRegistered values:")
print(df['Registered'].value_counts())

print("\nBounced values:")
print(df['Bounced'].value_counts())

print("\nTraffic Sources:")
print(df['Source'].value_counts())

print("\nLanding Pages:")
print(df['LandingPage'].value_counts())
# Remove duplicate sessions if any
print("Duplicate Session IDs:", df['SessionID'].duplicated().sum())

# Convert Date to datetime
df['Date'] = pd.to_datetime(df['Date'])

print("\nDate range:")
print(df['Date'].min(), "to", df['Date'].max())
total_sessions = len(df)

bounced_sessions = (df['Bounced'] == 'Yes').sum()

bounce_rate = (bounced_sessions / total_sessions) * 100

print("Total Sessions:", total_sessions)
print("Bounced Sessions:", bounced_sessions)
print(f"Bounce Rate: {bounce_rate:.2f}%")
avg_session_duration = df['SessionDurationSec'].mean()

print(f"Average Session Duration: {avg_session_duration:.2f} seconds")
avg_session_minutes = avg_session_duration / 60

print(f"Average Session Duration: {avg_session_minutes:.2f} minutes")
avg_pageviews = df['PageViews'].mean()

print(f"Average Pageviews per Session: {avg_pageviews:.2f}")
source_summary = df.groupby('Source').agg(
    Visits=('SessionID', 'count'),
    AvgSessionDuration=('SessionDurationSec', 'mean'),
    AvgPageViews=('PageViews', 'mean')
)

source_summary
source_conversion = (
    df.groupby('Source')['Registered']
    .apply(lambda x: (x == 'Yes').mean() * 100)
    .sort_values(ascending=False)
)

print("Conversion Rate by Traffic Source:")
print(source_conversion)
plt.figure(figsize=(9, 5))

source_conversion.plot(kind='bar')

plt.title('Conversion Rate by Traffic Source')
plt.xlabel('Traffic Source')
plt.ylabel('Conversion Rate (%)')

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
total_visits = len(df)

started_registration = (
    df['StartedRegistration'] == 'Yes'
).sum()

registered = (
    df['Registered'] == 'Yes'
).sum()

print("Total Visits:", total_visits)
print("Started Registration:", started_registration)
print("Completed Registration:", registered)
registration_start_rate = (
    started_registration / total_visits
) * 100

registration_completion_rate = (
    registered / started_registration
) * 100

overall_conversion_rate = (
    registered / total_visits
) * 100

print(f"Registration Start Rate: {registration_start_rate:.2f}%")
print(f"Registration Completion Rate: {registration_completion_rate:.2f}%")
print(f"Overall Conversion Rate: {overall_conversion_rate:.2f}%")
funnel_labels = [
    'Total Visits',
    'Started Registration',
    'Registered'
]

funnel_values = [
    total_visits,
    started_registration,
    registered
]

plt.figure(figsize=(8, 5))

plt.bar(funnel_labels, funnel_values)

plt.title('Website Conversion Funnel')
plt.ylabel('Number of Sessions')

plt.tight_layout()
plt.show()
landing_bounce = (
    df.groupby('LandingPage')['Bounced']
    .apply(lambda x: (x == 'Yes').mean() * 100)
    .sort_values(ascending=False)
)

print("Bounce Rate by Landing Page:")
print(landing_bounce)
plt.figure(figsize=(10, 6))

landing_bounce.plot(kind='bar')

plt.title('Bounce Rate by Landing Page')
plt.xlabel('Landing Page')
plt.ylabel('Bounce Rate (%)')

plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
source_summary['ConversionRate'] = (
    df.groupby('Source')['Registered']
    .apply(lambda x: (x == 'Yes').mean() * 100)
)

source_summary
plt.figure(figsize=(10, 6))

source_summary['ConversionRate'].sort_values(
    ascending=False
).plot(kind='bar')

plt.title('Traffic Source Conversion Performance')
plt.xlabel('Traffic Source')
plt.ylabel('Conversion Rate (%)')

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='Source',
    y='SessionDurationSec'
)

plt.title('Session Duration by Traffic Source')
plt.xlabel('Traffic Source')
plt.ylabel('Session Duration (seconds)')

plt.tight_layout()
plt.show()
kpi_summary = pd.DataFrame({
    'KPI': [
        'Total Visits',
        'Bounced Sessions',
        'Bounce Rate',
        'Average Session Duration (sec)',
        'Average Pageviews',
        'Started Registration',
        'Registered Users',
        'Registration Start Rate',
        'Registration Completion Rate',
        'Overall Conversion Rate'
    ],
    'Value': [
        total_visits,
        bounced_sessions,
        f'{bounce_rate:.2f}%',
        f'{avg_session_duration:.2f}',
        f'{avg_pageviews:.2f}',
        started_registration,
        registered,
        f'{registration_start_rate:.2f}%',
        f'{registration_completion_rate:.2f}%',
        f'{overall_conversion_rate:.2f}%'
    ]
})

kpi_summary
source_final = df.groupby('Source').agg(
    Visits=('SessionID', 'count'),
    Avg_Session_Duration=('SessionDurationSec', 'mean'),
    Avg_Pageviews=('PageViews', 'mean'),
    Conversions=('Registered', lambda x: (x == 'Yes').sum()),
    Conversion_Rate=('Registered', lambda x: (x == 'Yes').mean() * 100),
    Bounce_Rate=('Bounced', lambda x: (x == 'Yes').mean() * 100)
).round(2)

source_final = source_final.sort_values(
    'Conversion_Rate',
    ascending=False
)

source_final
page_final = df.groupby('LandingPage').agg(
    Visits=('SessionID', 'count'),
    Bounce_Rate=('Bounced', lambda x: (x == 'Yes').mean() * 100),
    Conversion_Rate=('Registered', lambda x: (x == 'Yes').mean() * 100)
).round(2)

page_final = page_final.sort_values(
    'Bounce_Rate',
    ascending=False
)

page_final
top_dropoff_pages = page_final.head(5)

top_dropoff_pages
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Traffic source conversion
source_final['Conversion_Rate'].sort_values().plot(
    kind='barh',
    ax=axes[0]
)

axes[0].set_title('Conversion Rate by Traffic Source')
axes[0].set_xlabel('Conversion Rate (%)')
axes[0].set_ylabel('Traffic Source')

# Landing page bounce rate
top_dropoff_pages['Bounce_Rate'].sort_values().plot(
    kind='barh',
    ax=axes[1]
)

axes[1].set_title('Top Landing Page Drop-off Points')
axes[1].set_xlabel('Bounce Rate (%)')
axes[1].set_ylabel('Landing Page')

plt.tight_layout()
plt.show()
print("========== WEB TRAFFIC KPI SUMMARY ==========")

print(f"Total Visits: {total_visits}")
print(f"Bounced Sessions: {bounced_sessions}")
print(f"Bounce Rate: {bounce_rate:.2f}%")
print(f"Average Session Duration: {avg_session_duration:.2f} seconds")
print(f"Average Pageviews: {avg_pageviews:.2f}")
print(f"Started Registration: {started_registration}")
print(f"Registered Users: {registered}")
print(f"Registration Start Rate: {registration_start_rate:.2f}%")
print(f"Registration Completion Rate: {registration_completion_rate:.2f}%")
print(f"Overall Conversion Rate: {overall_conversion_rate:.2f}%")
print("========== TRAFFIC SOURCE PERFORMANCE ==========")
display(source_final)
print("========== TOP DROP-OFF LANDING PAGES ==========")
display(top_dropoff_pages)
```
